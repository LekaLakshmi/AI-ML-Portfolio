# Loading in Dataframe from files

```python
# Read coffee csv file
coffee=pd.read_csv("coffee.csv")
print(f"Top 3 rows of coffee folder:\n{coffee.head(3)} ")
```

# Accessing Data with Pandas

- .loc selects by row label.
- .iloc selects by row position (starting at 0)

```python
# get full dataframe
print(f"Dataset:\n{coffee} ")
print(f"\n Get random sample :\n{coffee.sample(3,random_state=42)} ")

# update values in dataframe
coffee.loc[3:,'Units Sold']=100
print(f"Updated Dataframe:\n{coffee} ")

# access the row/column
print(f"First row:\n{coffee.loc[0]}\nFirst three rows:\n{coffee.loc[[0,1,2]]} ")

# Loc uses label
print(f"Get Day and unit sold for fifth to 7the row:\n{coffee.loc[5:8,["Day","Units Sold"]]} ")

# iloc uses position, we are changing index value and then try to access 
coffee.index=coffee.Day
print(f"First 3 rows:\n{coffee.iloc[0:3]} ")
print(f"Get only Monday:\n{coffee.loc["Monday"]} ")

# To access the column 
print(f"Day column:\n{coffee["Day"]}")

# Sort the values
print(f"\n\n\nSorted Unit sold column:\n{coffee.sort_values(["Units Sold","Coffee Type"],ascending=False)} ")

# Iterate over dataframe with for loop
for index,row in coffee.iterrows():
    print(index)
    print(row['Units Sold'],"\n")
```

# Filterning Data

```python
# Read bios file
bios=pd.read_csv('bios.csv')

# Filter height greater than 215
print(f"\nHeight Greater than 215\n{bios.loc[bios['height_cm']>215,['name']]} ")

# Another way
print(bios[bios['height_cm']>215][['name','height_cm']])

# height greater than 215 and born country is usa
bios[(bios['height_cm']>215) & (bios['born_country']=='USA')]

# name contains keith and patrick
bios[bios['name'].str.contains('Keith|patrick',case=False)]

# Filter name start with keith and country as usa france and great britain
bios[(bios['born_country'].isin(['USA','GBR','FRA'])) & (bios['name'].str.startswith("Keith"))
]

# Get country as usa and city as seattle
bios.query('born_country=="USA" and born_city=="Seattle"')
```

# Adding and removing column

```python
# Create a copy of the dataframe
coffee_new=coffee.copy()
coffee_new

# add new column price
coffee_new['price']=4.99
print(f"created new column:\n{coffee_new} ")

# Modify the price as per coffee type
import numpy as np
coffee_new['new price']=np.where(coffee_new['Coffee Type']=='Espresso',3.99,5.99)
print(f"Updated new price:\n{coffee_new} ")

# Drop the row 
coffee.drop('Monday')

# Drop the column
print(f"Dropped the column price:\n{coffee_new.drop(columns=['price'])} ")

# Add inplace to make update
coffee_new.drop(columns=['price'],inplace=True)

# just take only column that we need instead of using drop 
coffee_new=coffee_new[['Day','Units Sold','new price']]

# create new column by computation
coffee_new['Revenue']=coffee_new['Units Sold']*coffee_new['new price']

# create new column by computation
coffee_new['Revenue']=coffee_new['Units Sold']*coffee_new['new price']
coffee_new.head(3)

# Rename the column
coffee_new.rename(columns={'new price':'price'},inplace=True)


```

```python
# split the name into first name and find only for keith first name
bios_new=bios.copy()
bios_new['first_name']=bios_new['name'].str.split(' ').str[0]
print(bios_new.query('first_name=="Keith"'))
```

```python
# Date time
bios_new['born_date_time']=pd.to_datetime(bios_new['born_date'],format="%Y-%m-%d")
bios_new.info()
```

```python
bios_new['Born_year']=bios_new['born_date_time'].dt.year
bios_new[['name','Born_year']]
```

```python
bios_new['Height_Category']=bios_new['height_cm'].apply(lambda x: "Short" if x<165 else('Average' if x<185 else 'Tall'))
bios_new.head(3)
```

```python
def categorize_athlete(row):
    if row['height_cm']<175 and row['weight_kg']<70:
        return 'LightWeight'
    elif row['height_cm']<185 or row['weight_kg']<=80:
        return 'MiddleWeight'
    else:
        return 'HeavyWeight'
bios_new["category"]=bios_new.apply(categorize_athlete,axis=1)
bios_new.head(2)
```

# Handling Null values

```python
coffee=pd.read_csv('coffee.csv')
coffee.loc[[0,1],'Units Sold']=15
coffee.loc[[2,3],'Units Sold']=np.nan



# ways to fill NAN
coffee=coffee.fillna(1000)
coffee.fillna(coffee['Units Sold'].mean())

coffee=pd.read_csv('coffee.csv')
coffee.loc[[0,1],'Units Sold']=15
coffee.loc[[2,3],'Units Sold']=np.nan
coffee['Units Sold']=coffee['Units Sold'].interpolate()
coffee

# Drop rows with nan
coffee.loc[[0,1],'Units Sold']=15
coffee.loc[[2,3],'Units Sold']=np.nan
coffee.dropna(subset=['Units Sold'],inplace=True)

# Get rows that has NAN and does not has NAN
coffee=pd.read_csv('coffee.csv')
coffee.loc[[0,1],'Units Sold']=15
coffee.loc[[2,3],'Units Sold']=np.nan
coffee[coffee['Units Sold'].isna()]
coffee[coffee['Units Sold'].notna()]
```
