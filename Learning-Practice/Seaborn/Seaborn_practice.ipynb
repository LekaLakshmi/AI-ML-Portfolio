# Seaborn
- A Python library for creating data visualizations.
- Built on top of Matplotlib.
- Makes it easier to create attractive charts.

```python
# Import Seaborn for plotting and pandas for working with tables.
import seaborn as sns
import pandas as pd
```

```python
# Check which Seaborn version is installed.
print(f"Version of seaborn: {sns.__version__} ")
```

```python
# list down dataset availble in seaborn
print(f"List of dataset:\n{sns.get_dataset_names()} ")

# Read flight dataset as dataframe
fight_dataframe=sns.load_dataset('flights')

# print flight dataset
print(f"\n\nTop 5 rows of flight dataset:\n{fight_dataframe.head()}" )
```

# 2. Distribution plots
### Histograms show how values are spread across ranges. 

```python
# Take passenger column
passengers=fight_dataframe['passengers']

# Plot histogram
sns.displot(passengers)
```

```python
# Load the car-crashes sample dataset.
crashes_dataframe=sns.load_dataset('car_crashes')
crashes_dataframe.head()
```

```python
# Older API: distplot combines a histogram and a smooth density curve; prefer histplot in new notebooks.
# Take total column and plot histogram
sns.distplot(crashes_dataframe['total'],bins=25)
```

```python
# displot displays the distribution of total crash values as a histogram.
# Take total column and plot histogram
sns.displot(crashes_dataframe['total'],bins=25)
```

# 3. Relationships and pairwise plots
### Joint plots compare two variables. Pair plots compare several numeric variables at once.

```python
# Compare alcohol and speeding values; the side plots show each variable's distribution.
# correlation and distribution 
sns.jointplot(x='alcohol',y='speeding',data=crashes_dataframe)
```

```python
# Add a fitted regression line to the joint plot.
# Get regression line
sns.jointplot(x='alcohol',y='speeding',data=crashes_dataframe,kind='reg')
```

```python
# Show a smoothed two-variable density estimate with KDE contours.
sns.jointplot(x='alcohol',y='speeding',data=crashes_dataframe,kind='kde')
```

```python
# Use hexagons to show where pairs of observations are concentrated.
sns.jointplot(x='alcohol',y='speeding',data=crashes_dataframe,kind='hex')
```

```python
# Plot a smoothed density curve for speeding values.
sns.kdeplot(crashes_dataframe['speeding'])
```

```python
# Compare pairwise relationships among the dataset's variables.
sns.pairplot(crashes_dataframe)
```

```python
# Load the tips sample dataset and preview its rows before plotting.
tips_dataframe=sns.load_dataset('tips')
print(tips_dataframe.head())

# use pairplots
sns.pairplot(tips_dataframe)
```

```python
# Color pairwise plots by sex to compare the two groups.
sns.pairplot(tips_dataframe,hue='sex')
```

# 4. Rug plot
### A rug plot places a small mark for each observation along an axis. Marks that gather together show where values are more common.

- Histogram - which is tall is more similar data points
- rugs - which are very near more similar data points

```python
# Each rug mark represents one total-bill observation.
sns.rugplot(tips_dataframe['total_bill'])
```

# 5. Plot appearance
### Seaborn styles change backgrounds, grid lines, and axes. Each set_style call replaces the previously selected style. Context settings change text and line sizes for different display settings.

```python
# White style: a clean white background with no grid by default.
sns.set_style('white')
sns.jointplot(x='alcohol',y='speeding',data=crashes_dataframe)
```

```python
# Whitegrid style: white background with grid lines.
sns.set_style('whitegrid')
sns.jointplot(x='alcohol',y='speeding',data=crashes_dataframe)
```

```python
# Dark style: light gray plotting background without grid lines.
sns.set_style('dark')
sns.jointplot(x='alcohol',y='speeding',data=crashes_dataframe)
```

```python
# Darkgrid style: light gray plotting background with grid lines.
sns.set_style('darkgrid')
sns.jointplot(x='alcohol',y='speeding',data=crashes_dataframe)
```

```python
# Ticks style emphasizes axis tick marks.
sns.set_style('ticks')
sns.jointplot(x='alcohol',y='speeding',data=crashes_dataframe)
```

```python
# Resize the plot
import matplotlib.pyplot as plt
plt.figure(figsize=(6,3))
sns.jointplot(x='alcohol',y='speeding',data=crashes_dataframe)
```

```python
# Paper context uses smaller elements suited to a compact figure.
sns.set_context('paper',font_scale=1.5)
sns.jointplot(x='alcohol',y='speeding',data=crashes_dataframe)
```

```python
# Talk context uses larger elements for presentations.
sns.set_context('talk')
sns.jointplot(x='alcohol',y='speeding',data=crashes_dataframe)
```

```python
# Poster context uses even larger text and plot elements.
sns.set_context('poster')
sns.jointplot(x='alcohol',y='speeding',data=crashes_dataframe)
```

```python
# Hide the bottom spine (the bottom border line) of the joint plot.
sns.jointplot(x='alcohol',y='speeding',data=crashes_dataframe)
sns.despine(bottom=True)
```

# 6. Categorical plots
### These plots summarize or show observations grouped by categories such as day or sex. 
- Bar plots show a summary statistic
- count plots show category counts
- box and violin plots show distributions
- swarm and strip plots show individual observations.

```python
# A barplot summarizes total bills by sex; here the estimator is the median.
import numpy as np
sns.barplot(x='sex',y='total_bill',data=tips_dataframe,estimator=np.median) # we can change to standard deviation, variance, mean
```

```python
# Count how many rows belong to each sex category.
sns.countplot(x='sex',data=tips_dataframe)
```

```python
# Preview the tips table before making categorical plots.
tips_dataframe.head()
```

```python
# Compare bill distributions by day and sex with a boxplot.
import matplotlib.pyplot as plt
sns.set_context('paper')
plt.figure()
sns.boxplot(x="day", y="total_bill", data=tips_dataframe,hue='sex')
# move the legend
plt.legend(loc=0)
```

```python
# Split each violin by sex; wider areas represent more observations near that value.
sns.violinplot(x="day", y="total_bill", data=tips_dataframe,hue='sex',split=True)
```

```python
# Overlay individual bills on the split violins to show the actual observations.
sns.violinplot(x="day", y="total_bill", data=tips_dataframe,hue='sex',split=True)
sns.swarmplot(x="day", y="total_bill", data=tips_dataframe,hue='sex',color='white')
plt.legend(loc=0)
```

```python
# A swarm plot moves dots sideways to reduce overlap.
sns.swarmplot(x="day", y="total_bill", data=tips_dataframe,hue='sex')
plt.legend(loc=0)
```

```python
# A strip plot jitters dots sideways; dodge separates hue groups within each day.
sns.stripplot(x='day',y='total_bill',data=tips_dataframe,jitter=True,hue='sex',dodge=True)
plt.legend(loc=0)
```

```python
# Try the Greens color palette.
sns.stripplot(x='day',y='total_bill',data=tips_dataframe,jitter=True,hue='sex',dodge=True,palette='Greens')
```

# 7. Heatmaps and clustered heatmaps
- A heatmap uses color to show values in a table. Here, the numeric crash columns are correlated, and a second heatmap displays passenger counts by year and month.
- A clustermap also groups similar rows and columns.

```python
# Calculate correlations using only numeric columns.
crashes_dataframe.head()
crashed_correlation=crashes_dataframe.select_dtypes(include=['number']).corr()
crashed_correlation.head()
```

```python
# Annotate each heatmap cell with its correlation value.
sns.heatmap(crashed_correlation,cmap='Blues',annot=True)
```

```python
# Load the flights dataset for a year-by-month table.
flights_df=sns.load_dataset('flights')
flights_df.head()
```

```python
# Reshape passenger counts so years are rows and months are columns, then plot them.
flights=flights_df.pivot_table(index='year',columns='month',values='passengers')
sns.heatmap(flights,cmap='hot')
```

```python
# Cluster rows and columns based on similar passenger-count patterns.
sns.clustermap(flights,standard_scale=1,cmap='Greens')
```

# 8. PairGrid customization
-  PairGrid lets you choose what appears in different parts of a pairwise plot.
-  map_diag draws on the diagonal, while map_upper and map_lower draw above and below it. 
- Create a fresh grid before remapping so old plots do not remain underneath.
- pairgrid we can do many customization

```python
# pairplot is convenient; PairGrid provides more control over each region.
sns.pairplot(tips_dataframe,hue='sex') 
```

```python
# Load iris data and make the standard pairplot.
iris=sns.load_dataset('iris')
sns.pairplot(iris)
```

```python
# Map scatter plots to every cell of the PairGrid, including the diagonal.
iris_grid=sns.PairGrid(iris)
iris_grid.map(plt.scatter) # this is similar to pairplot
```

```python
# histograms are added over existing diagonal scatter plots.
plt.figure()
iris_grid.map(plt.scatter)
iris_grid.map_diag(plt.hist) # get diagonla histogram
```

```python
# scatter plots off-diagonal and histograms on the diagonal.
plt.figure()
iris_grid=sns.PairGrid(iris)
iris_grid.map_offdiag(plt.scatter)
iris_grid.map_diag(plt.hist)
```

```python
# Use KDE plots above the diagonal, scatter plots below, and histograms on the diagonal.
iris_grid=sns.PairGrid(iris)
iris_grid.map_diag(plt.hist)
iris_grid.map_upper(sns.kdeplot)
iris_grid.map_lower(plt.scatter)
```

```python
# Limit the grid to selected x/y columns, color by species, 
iris=sns.load_dataset('iris')
iris_grid=sns.PairGrid(iris,hue='species',x_vars=["petal_length","sepal_length"],y_vars=["petal_width","sepal_width"])
iris_grid.map(sns.scatterplot)
iris_grid.add_legend()
```

# 9. Regression plots
- lmplot shows the relationship between two numeric variables and fits a regression line.
-  hue separates groups, and markers chooses the point symbols for those groups.

```python
# Compare total bill with tip and fit a separate regression line for each sex group.
sns.lmplot(x='total_bill',y='tip',data=tips_dataframe,hue='sex',markers=['*',"<"],scatter_kws={"s":30,'edgecolor':'b'})
```
