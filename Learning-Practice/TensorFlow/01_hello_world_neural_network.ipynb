# Hello World: Learning with a Neural Network

## Objective
Train a simple TensorFlow/Keras model to learn the relationship **y = 2x − 1** from example data. 
Then use it to predict the output for a new input, x = 10.

```python
import tensorflow as tf
import numpy as np
```

## 1. Prepare the Data

The model learns from examples of input values (`xs`) and their
corresponding output values (`ys`). In this example, the values follow
the rule **y = 2x − 1**.

```python
# Declare model inputs and outputs for training
xs = np.array([-1.0,  0.0, 1.0, 2.0, 3.0, 4.0], dtype=float)
ys = np.array([-3.0, -1.0, 1.0, 3.0, 5.0, 7.0], dtype=float)
```

## 2. Build and Compile the Model

This model has one Dense layer with one unit. For a single input and
output, the layer learns a weight and a bias, giving it the form
**y = weight × x + bias**.

The loss function measures prediction error. The SGD optimizer adjusts
the model’s weight and bias to reduce that error.

```python
# Build model
model=tf.keras.Sequential(
    [tf.keras.Input(shape=(1,)), # input shape
     tf.keras.layers.Dense(units=1) # one unit layer
    ]
)

# Compile model
model.compile(optimizer='sgd',loss='mean_squared_error')

```

## 3. Train the Model

`model.fit()` trains the model using the examples. An epoch is one
pass through the training data. During training, the model updates its
weight and bias to reduce the loss.

```python
# train model
model.fit(xs,ys,epochs=1000,verbose=0)
```

## 4. Predict for a New Input

The model has not trained on `x = 10`. Use `model.predict()` to ask it
for an output for this new input.

```python
# Model prediction
prediction = model.predict(np.array([[10.0]]), verbose=0)[0][0]
print(f"Prediction for x=10: {prediction:.6f}")
print(f"Difference from 19: {abs(prediction - 19):.6f}")
```

## Result

The expected output is **19**, because **2 × 10 − 1 = 19**. The model’s
prediction may be slightly different because training adjusts its
weight and bias iteratively and may not reach the exact values 2 and −1.

This is a numeric prediction task (regression). The model outputs a
number; it does not automatically output a probability.
