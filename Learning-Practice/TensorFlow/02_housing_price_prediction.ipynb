# Housing Price Prediction with TensorFlow

## Objective
Train a simple neural network to learn how the number of bedrooms relates
to a house price, then predict the price for a house with 7 bedrooms.

This is a small learning example, not a real-world housing-price model.

```python
import tensorflow as tf
import numpy as np
```

## 1. Prepare the Data

The data contains six examples. Each input is the number of bedrooms, and
each target is the house price in hundreds of thousands of dollars.

```python
# create function to create train data
def create_training_data():
    # array for bedrooms
    number_bedrooms=np.array([1,2,3,4,5,6],dtype=float)
    # array for house price
    house_price=np.array([1.0,1.5,2.0,2.5,3.0,3.5],dtype=float)
    # return the training data
    return number_bedrooms,house_price

# Call create training function
features,target=create_training_data()
# Get shape of features and label
print(f"Feature shape :{features.shape}\nLabel shape: {target.shape} ")
```

## 2. Build and Compile the Model

The model has one Dense layer with one unit. It learns a weight and a bias
to connect the number of bedrooms to the predicted price.

The loss measures prediction error. The SGD optimizer adjusts the model’s
weight and bias to reduce that error.

```python
# define and compile model function
def create_compile_model():
    # define model | input shape and layer 
    model=tf.keras.Sequential([
        tf.keras.Input(shape=(1,)),
        tf.keras.layers.Dense(units=1)
    ])
    # compile model
    model.compile(optimizer='sgd',loss='mean_squared_error')
    # return model
    return model

# Call create compile function and see the summary
untrained_model=create_compile_model()
untrained_model.summary()
```

## 3. Train the Model

`model.fit()` trains the model using the bedroom and price examples.
An epoch is one pass through the training data.

```python
# train model function
def train_model():
    # create dataset
    features,target=create_training_data()
    # create model
    model=create_compile_model()
    # train model
    model.fit(features,target,epochs=1000,verbose=0)
    # return model
    return model

# call train model function
trained_model=train_model()
```


## 4. Predict a Price

Ask the trained model to predict the price for a house with 7 bedrooms.


```python
# Make new prediction
new_number_bedrooms = np.array([[7.0]])
predicted_price = trained_model.predict(new_number_bedrooms, verbose=0).item()

print(
    f"Predicted price: {predicted_price:.2f} hundreds of thousands "
    f"of dollars for 7 bedrooms."
)
```

## Result

The model returns its predicted price in hundreds of thousands of dollars.
For example, a prediction of 4.0 represents $400,000.

This is a simple demonstration using only bedroom count. Real housing
prices depend on many other factors, so this model should not be used
to estimate actual market prices.
